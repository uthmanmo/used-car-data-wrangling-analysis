```python
import pandas as pd

 file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-SkillsNetwork/labs/Data%20files/auto.csv"

 headers = ["symboling","normalized-losses","make","fuel-type","aspiration",
           "num-of-doors","body-style","drive-wheels","engine-location",
           "wheel-base", "length","width","height","curb-weight","engine-type",
           "num-of-cylinders", "engine-size","fuel-system","bore","stroke",
           "compression-ratio","horsepower","peak-rpm","city-mpg","highway-mpg","price"]

 df = pd.read_csv(file_path, names=headers, header=None)

 print(df.head())
```

```python
import numpy as np

df.replace("?", np.nan, inplace = True)
df.head(5)
```

```python
avg_norm_loss = df["normalized-losses"].astype("float").mean(axis=0)
print("Average of normalized-losses:", avg_norm_loss)
```

```python
df["normalized-losses"] = df["normalized-losses"].replace(np.nan, avg_norm_loss)
```

```python
avg_bore=df['bore'].astype('float').mean(axis=0)
print("Average of bore:", avg_bore)
```

```python
df["bore"] = df["bore"].replace(np.nan, avg_bore)
```

```python
avg_stroke = df["stroke"].astype("float").mean(axis=0)
print("Average of stroke:", avg_stroke)
```

```python
df["stroke"] = df["stroke"].replace(np.nan, avg_stroke)
```

```python
avg_horsepower = df['horsepower'].astype('float').mean(axis=0)
print("Average horsepower:", avg_horsepower)
```

```python
df['horsepower'] = df['horsepower'].replace(np.nan, avg_horsepower)
```

```python
avg_peakrpm=df['peak-rpm'].astype('float').mean(axis=0)
print("Average peak rpm:", avg_peakrpm)
```

```python
df['peak-rpm'] = df['peak-rpm'].replace(np.nan, avg_peakrpm)
```

```python
df['num-of-doors'].value_counts()
```

```python
df['num-of-doors'].value_counts().idxmax
```

```python
df.replace({"num-of-doors": {np.nan: "four"}}, inplace=True)
```

```python
df.dropna(subset=["price"], axis=0, inplace=True)

df.reset_index(drop=True, inplace=True)
```

```python
df[["bore", "stroke"]] = df[["bore", "stroke"]].astype("float")
df[["normalized-losses"]] = df[["normalized-losses"]].astype("int")
df[["price"]] = df[["price"]].astype("float")
df[["peak-rpm"]] = df[["peak-rpm"]].astype("float")
```

```python
df["horsepower"] = df["horsepower"].astype(int, copy=True)

print("Min horsepower:", min(df["horsepower"]))
print("Max horsepower:", max(df["horsepower"]))

if min(df["horsepower"]) == max(df["horsepower"]):
    print("All horsepower values are identical. Using one bin.")
    single_value = min(df["horsepower"])
    
    bins = [single_value - 0.5, single_value + 0.5]
    group_names = ['Medium']  # Just one label for one bin
    
    df['horsepower-binned'] = 'Medium'  
else:
    min_hp = min(df["horsepower"])
    max_hp = max(df["horsepower"])
    step = (max_hp - min_hp) / 3
    
    bins = [min_hp, min_hp + step, min_hp + 2*step, max_hp]
    group_names = ['Low', 'Medium', 'High']
    
    df['horsepower-binned'] = pd.cut(df['horsepower'], bins, labels=group_names, include_lowest=True)

print("Bins:", bins)
print(df[['horsepower', 'horsepower-binned']].head(20))

print("\nDistribution of values in bins:")
print(df["horsepower-binned"].value_counts())

%matplotlib inline
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
df["horsepower-binned"].value_counts().plot(kind='bar')
plt.xlabel("Horsepower Category")
plt.ylabel("Count")
plt.title("Horsepower Bins")
plt.show()

plt.figure(figsize=(8, 5))
if min(df["horsepower"]) == max(df["horsepower"]):
    plt.hist([min(df["horsepower"])] * len(df["horsepower"]), bins=1)
else:
    plt.hist(df["horsepower"], bins=3)
plt.xlabel("Horsepower")
plt.ylabel("Count")
plt.title("Horsepower Distribution")
plt.show()
```
